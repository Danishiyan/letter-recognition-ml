# 01 · Exploratory Data Analysis (EDA)

**Goal:** understand the UCI Letter Recognition dataset *before* building any model.

Questions this notebook answers:
1. How big is the dataset, and are there missing values?
2. Are the 26 letters balanced (similar counts)?

## 1. Setup & load the data
We add the project root to Python's search path so we can import our own `src` package.

```python
import sys
from pathlib import Path

# notebooks/ is one level below the project root
PROJECT_ROOT = Path.cwd().parent
sys.path.append(str(PROJECT_ROOT))

import matplotlib.pyplot as plt
import seaborn as sns

from src.data import load_data

df = load_data()
print("Shape:", df.shape)
df.head()
```

## 2. Overview: types and missing values

```python
df.info()
print("\nTotal missing values:", df.isna().sum().sum())
```

## 3. Class balance: how many rows per letter?
If some letters were much rarer, a model could just ignore them and still score well. So we check before modelling.

```python
counts = df["letter"].value_counts().sort_index()

fig, ax = plt.subplots(figsize=(12, 4))
sns.barplot(x=counts.index, y=counts.values, color="steelblue", ax=ax)
ax.axhline(counts.mean(), color="red", linestyle="--", label=f"mean = {counts.mean():.0f}")
ax.set(title="Number of samples per letter", xlabel="Letter", ylabel="Count")
ax.legend()
plt.tight_layout()
plt.savefig(PROJECT_ROOT / "reports" / "figures" / "class_distribution.png", dpi=150)
plt.show()

print("Min:", counts.min(), "(" + counts.idxmin() + ")", "| Max:", counts.max(), "(" + counts.idxmax() + ")")
```
